# Câu 1

```python
 #1: C.144 ngày

orders = pd.read_csv("orders.csv", parse_dates=['order_date'])

# sort
orders = orders.sort_values(['customer_id', 'order_date'])

# tính khoảng cách ngày giữa các đơn
orders['prev_date'] = orders.groupby('customer_id')['order_date'].shift(1)
orders['gap_days'] = (orders['order_date'] - orders['prev_date']).dt.days

# chỉ lấy khách có >1 đơn
gap = orders['gap_days'].dropna()

median_gap = gap.median()
print("Median gap:", median_gap)
```

# Câu 2

```python
#2: D. Standard

products = pd.read_csv("products.csv")

# tính margin
products['margin'] = (products['price'] - products['cogs']) / products['price']

# group theo segment
segment_margin = products.groupby('segment')['margin'].mean().sort_values(ascending=False)

print(segment_margin)
print("Best segment:", segment_margin.idxmax())
```

# Câu 6

```python
# Q6: A. 55+

import pandas as pd

customers = pd.read_csv("customers.csv")
orders = pd.read_csv("orders.csv")

# đếm số đơn của từng customer
order_counts = orders.groupby("customer_id").size().reset_index(name="num_orders")

# join với customers
df = customers.merge(order_counts, on="customer_id", how="left")

# khách không có đơn thì num_orders = 0
df["num_orders"] = df["num_orders"].fillna(0)

# lọc age_group khác null
df_age = df[df["age_group"].notna()]

# tính trung bình số đơn / khách hàng theo age_group
avg_orders = df_age.groupby("age_group")["num_orders"].mean().sort_values(ascending=False)

print(avg_orders)
print("Top age_group:", avg_orders.idxmax())
```

# Câu 7

```python
# Q7: C. East

orders = pd.read_csv("orders.csv")
order_items = pd.read_csv("order_items.csv")
geography = pd.read_csv("geography.csv")

order_items['revenue'] = (
    order_items['quantity'] * order_items['unit_price']
    - order_items['discount_amount']
)

df = order_items.merge(orders[['order_id', 'zip']], on='order_id', how='left')
df = df.merge(geography[['zip', 'region']], on='zip', how='left')

region_revenue = df.groupby('region')['revenue'].sum().reset_index()

region_revenue = region_revenue.sort_values(by='revenue', ascending=False)
print(region_revenue)
top_region = region_revenue.head(1)

print("\n Best Region")
print(top_region)
```

```python
# Q8: A. credit_card

orders = pd.read_csv("orders.csv")
payments = pd.read_csv("payments.csv")

cancelled = orders[orders['order_status'] == 'cancelled']

df = cancelled.merge(payments, on='order_id')

result = df['payment_method_y'].value_counts()

print(result)
print("Top:", result.idxmax())
```

# Câu 9

```python
# Q9: A. S

# Load data
products = pd.read_csv("products.csv")
order_items = pd.read_csv("order_items.csv")
returns = pd.read_csv("returns.csv")

# Gắn size vào order_items
order_with_size = pd.merge(order_items, products[['product_id', 'size']], on='product_id', how='left')

# Tổng số item theo size
total_by_size = order_with_size.groupby('size')['order_id'].count()

# Lọc các item bị trả hàng
returned_items = pd.merge(returns, order_with_size, on=['order_id', 'product_id'], how='left')

# Đếm số item bị trả theo size
returned_by_size = returned_items.groupby('size')['order_id'].count()

# eturn rate
return_rate = (returned_by_size / total_by_size).sort_values(ascending=False)

print(return_rate)
print("Highest:", return_rate.index[0])
```
